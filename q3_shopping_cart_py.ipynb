```python
#Part C

def create_cart(owner,discount=0):
  return {"owner":owner,
          "discount":discount,
          "items":[]}

def add_to_cart(cart,name,price,qty=1):
  cart["items"].append({
      "name":name,
      "price":price,
      "qty":qty})
  return cart

def update_price(price_tuple,new_price):
  try:
     price_tuple[1]=new_price
  except TypeError :
    print("Tuple is immutable")
  #tuple are immutable

def calculate_total(cart):
  total=0
  for item in cart["items"]:
    total+=item["price"]*item["qty"]
  return total-total*cart["discount"]/100


cart1 = create_cart("Aarav", 10)

add_to_cart(cart1, "Laptop", 50000, 1)
add_to_cart(cart1, "Mouse", 500, 2)

# Customer 2
cart2 = create_cart("Priya", 5)

add_to_cart(cart2, "Book", 300, 3)
add_to_cart(cart2, "Pen", 20, 5)

# Display carts
print("Customer 1 Cart:")
print(cart1)

print("\nCustomer 2 Cart:")
print(cart2)

# Calculate totals
print("\nCustomer 1 Total:", calculate_total(cart1))
print("Customer 2 Total:", calculate_total(cart2))

# Tuple demonstration
price_info = (1000, "Laptop")
update_price(price_info, 1200)
```
